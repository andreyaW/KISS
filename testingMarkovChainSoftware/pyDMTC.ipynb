```python
def plot_sequence(mc, sequence):

    import matplotlib.pyplot as plt

    plt.figure(figsize=(12, 6))
    plt.plot(sequence, marker='o')
    plt.yticks(range(len(mc.states)), mc.states)
    plt.xlabel('Step')
    plt.ylabel('State')
    plt.gca().invert_yaxis() # invert y-axis to have the first state at the top
    plt.title('Markov Chain State Sequence')
    plt.grid()
    plt.show()
```

```python
# Example 1: Discrete-Time Markov Chain using PyDTMC
from pydtmc import MarkovChain

import pydtmc
import numpy as np
import matplotlib.pyplot as plt

# Define transition matrix (rows sum to 1)
# States: [Healthy, Degraded, Failed]
P = np.array([
    [0.90, 0.09, 0.01],   # Healthy can stay, degrade, or fail
    [0.00, 0.85, 0.15],   # Degraded can stay or fail
    [0.10, 0.00, 0.90]    # Failed may be repaired (10%) or remain failed
])

states = ['Healthy', 'Degraded', 'Failed']

# Build Markov Chain
mc = MarkovChain(P, states)

# simulate the Markov Chain for 100 steps starting from 'Healthy'
initial_state = 'Healthy'
sequence = mc.simulate(100, initial_state, output_indices=True)

print("Steady-state distribution:")
print('π:', list(mc.steady_states))   # πP = π

# plot the Markov Chain transition diagram
pydtmc.plot_graph(mc)
plot_sequence(mc, sequence)
# pydtmc.plot_flow(mc, steps=10, interval=2)
```
