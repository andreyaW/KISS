```python
# Example 2: Hidden Markov Model using ChainoPy
from chainopy import HMM
import numpy as np
import matplotlib.pyplot as plt

# Hidden states
states = ['Healthy', 'Degraded', 'Failed']

# Transition matrix for hidden states (similar to above)
A = np.array([
    [0.90, 0.09, 0.01],
    [0.00, 0.85, 0.15],
    [0.10, 0.00, 0.90]
])

# Emission probabilities (sensor readings)
# Rows correspond to hidden states, columns to observed symbols
observations = ['Low Vib', 'Medium Vib', 'High Vib']
B = np.array([
    [0.80, 0.18, 0.02],  # Healthy mostly emits Low Vib
    [0.10, 0.70, 0.20],  # Degraded emits Medium Vib most often
    [0.05, 0.15, 0.80]   # Failed emits High Vib most often
])

# Initial probabilities
pi = np.array([1.0, 0.0, 0.0])  # Start Healthy

# Build HMM
model = HMM(states, observations, A, B, pi)

# Simulate 40 time steps of hidden + observed data
hidden_seq, observed_seq = model.simulate(40)
print("True hidden states:\n", hidden_seq)
print("Observed emissions:\n", observed_seq)

# Decode most likely hidden states from observed emissions
decoded_seq = model.viterbi(observed_seq)
print("\nViterbi-decoded states:\n", decoded_seq)

# Plot the hidden vs decoded
plt.figure(figsize=(10, 3))
plt.plot(hidden_seq, 'o-', label='True hidden state')
plt.plot(decoded_seq, 'x--', label='Viterbi decoded')
plt.xticks(range(len(hidden_seq)))
plt.legend()
plt.title('Hidden vs Decoded States (ChainoPy HMM)')
plt.xlabel('Time step')
plt.show()
```
