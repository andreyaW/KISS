adding sensors to the seriesComps obj and printing output to excel 

```python
from shipClass.Component import Component
from shipClass.SeriesComps import SeriesComps
from shipClass.System import System
from shipClass.SensedSystem import SensedSystem
# from shipClass.Sensor_basic import Sensor

# create a seriesComps object with 3 components in series
comps = [Component(f'comp_{i}', MTTF=20, MTTR='NR') for i in range(3)]
series_comps = SeriesComps(comps)

# create a sensed system with 5 sensors on each component
system = System('main system' , [series_comps])
sensed_system = SensedSystem(system, number_of_sensors=[5])
sensed_system.simulate(100)
sensed_system.plotHistory()   
sensed_system.printHistory2Excel('testResults/series_comps_test.xlsx')
```
