```python
from sympy import symbols, diff

x, y = symbols('x y')

f = x**2 + y**2
dfdx = diff(f, x)
dfdy = diff(f, y)
print(f"f(x, y) = {f}")
print(f"∂f/∂x = {dfdx}")
print(f"∂f/∂y = {dfdy}")
```

z(t) = failure rate function
f(t) = probability density function
F(t) = cumulative distribution function
R(t) = survivor function = 1- F(t)


```python

from sympy import symbols, diff, oo, integrate

t = symbols('t')
R_t = 1 / ((0.2*t + 1)**2)
f_t = diff(R_t, t) *-1
z_t = f_t / R_t
MTTF = integrate(R_t, (t, 0, oo))


print(f"R(t) = {R_t}")
print(f"f(t) = {f_t}")
print(f"z(t) = {z_t}")
print(f"MTTF = {MTTF:.2f} months")
```
